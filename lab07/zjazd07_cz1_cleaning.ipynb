```python
df_clean = pd.DataFrame(data)

# KROK 1: Napraw wynagrodzenie
df_clean['wynagrodzenie'] = df_clean['wynagrodzenie'].replace('brak', np.nan)
df_clean['wynagrodzenie'] = pd.to_numeric(df_clean['wynagrodzenie'], errors='coerce')
print(f"Po kroku 1 — NaN w wynagrodzenie: {df_clean['wynagrodzenie'].isna().sum()}")  # 2

# KROK 2: Usuń duplikaty
df_clean = df_clean.drop_duplicates().reset_index(drop=True)
print(f"Po kroku 2 — Shape: {df_clean.shape}")  # (25, 6)

# KROK 3: Wypełnij NaN w wynagrodzeniu medianą
mediana_clean = df_clean['wynagrodzenie'].median()
print(f"Po kroku 3 — Mediana: {mediana_clean}")  # 5100.0
df_clean['wynagrodzenie'] = df_clean['wynagrodzenie'].fillna(mediana_clean)
print(f"Po kroku 3 — NaN w wynagrodzenie: {df_clean['wynagrodzenie'].isna().sum()}")  # 0

# KROK 4: Wypełnij NaN w ocena_roczna średnią
srednia_clean = round(df_clean['ocena_roczna'].mean(), 2)
print(f"Po kroku 4 — Średnia ocena: {srednia_clean}")  # 4.26
df_clean['ocena_roczna'] = df_clean['ocena_roczna'].fillna(srednia_clean)
print(f"Po kroku 4 — NaN w ocena_roczna: {df_clean['ocena_roczna'].isna().sum()}")  # 0

# KROK 5: Wyczyść imie
df_clean['imie'] = df_clean['imie'].str.strip().str.title()

# KROK 6: Wyczyść dzial
df_clean['dzial'] = df_clean['dzial'].str.strip().str.title()
df_clean['dzial'] = df_clean['dzial'].str.replace('Hr', 'HR', regex=False)
df_clean['dzial'] = df_clean['dzial'].str.replace('It', 'IT', regex=False)

# KROK 7: Konwersja daty
df_clean['data_zatrudnienia'] = pd.to_datetime(df_clean['data_zatrudnienia'], errors='coerce')

# WERYFIKACJA
print("\n=== WERYFIKACJA ===")
print(f"Shape: {df_clean.shape}")
print(f"\nDtypes:\n{df_clean.dtypes}")
print(f"\nNaN per kolumna:\n{df_clean.isna().sum()}")
print(f"\nUnikalne działy: {sorted(df_clean['dzial'].unique())}")
print(f"\nPierwsze imiona: {df_clean['imie'].tolist()[:8]}")
```

```python
# 1. Średnie wynagrodzenie per dział
print("Średnie wynagrodzenie per dział:")
print(df_clean.groupby('dzial')['wynagrodzenie'].mean().round(2))

# 2. Średnia ocena per dział
print("\nŚrednia ocena per dział:")
print(df_clean.groupby('dzial')['ocena_roczna'].mean().round(2))

# 3. Top 3 najlepiej opłacani
print("\nTop 3 wynagrodzenia:")
print(df_clean.nlargest(3, 'wynagrodzenie')[['imie', 'dzial', 'wynagrodzenie']])

# 4. Liczba pracowników per dział
print("\nLiczba pracowników per dział:")
print(df_clean['dzial'].value_counts())
```

```python
# 1. Najdłuższy staż (w latach)
najwczesniejsza = df_clean['data_zatrudnienia'].min()
dzis = pd.Timestamp('today')
staz_lat = (dzis - najwczesniejsza).days / 365.25
print(f"Najdłuższy staż: {staz_lat:.1f} lat")

# 2. Najwyższa ocena w każdym dziale
print("\nNajwyżej oceniani per dział:")
top_ocena = df_clean.loc[df_clean.groupby('dzial')['ocena_roczna'].idxmax(), ['dzial', 'imie', 'ocena_roczna']]
print(top_ocena)

# 3. Korelacja staż <-> wynagrodzenie
df_clean['staz_dni'] = (dzis - df_clean['data_zatrudnienia']).dt.days
korelacja = df_clean[['staz_dni', 'wynagrodzenie']].corr().iloc[0, 1]
print(f"\nKorelacja staż-wynagrodzenie: {korelacja:.3f}")
```
