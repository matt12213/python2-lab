## Ćwiczenie 1: Series — tworzenie i operacje

```python
# Series z listy — przykład ze skryptu
oceny = pd.Series([4.5, 3.0, 5.0, 3.5, 4.0])
print(oceny)
```

### Zadanie 1a — Series `pensje` ze słownika

```python
pensje = pd.Series({
    'Anna':  5500,
    'Jan':   7200,
    'Ewa':   4800,
    'Marek': 9100,
    'Kasia': 6300
})
print(pensje)

# Sprawdzenie: pensje['Marek'] powinno być 9100
print("\npensje['Marek'] =", pensje['Marek'])
```

### Zadanie 1b — Odpowiedzi na pytania o `pensje`

```python
# 1. Imię osoby zarabiającej najwięcej
print("Osoba zarabiająca najwięcej:", pensje.idxmax())

# 2. Średnia pensja
srednia = pensje.mean()
print("Średnia pensja:", srednia)

# 3. Osoby zarabiające powyżej średniej
powyzej_sredniej = pensje[pensje > srednia]
print("\nOsoby zarabiające powyżej średniej:")
print(powyzej_sredniej)

# 4. Liczba osób zarabiających poniżej 6000
ponizej_6000 = (pensje < 6000).sum()
print("\nLiczba osób zarabiających poniżej 6000 zł:", ponizej_6000)
```

### Zadanie 1c — Operacje na Series (podwyżki, premie)

```python
# 1. Pensje po podwyżce 10%
pensje_podwyzka = pensje * 1.10
print("Pensje po podwyżce 10%:")
print(pensje_podwyzka)

# 2. Pensje po premii 500 zł
pensje_premia = pensje + 500
print("\nPensje po premii 500 zł:")
print(pensje_premia)

# 3. Nowa pensja = (pensja * 1.10) + 500
nowe_pensje = (pensje * 1.10) + 500
print("\nNowe pensje (podwyżka + premia):")
print(nowe_pensje)

# 4. Ranking od najwyższej
print("\nRanking nowych pensji (malejąco):")
print(nowe_pensje.sort_values(ascending=False))

# Sprawdzenie: nowa pensja Anny = 5500 * 1.10 + 500 = 6550.0
print("\nNowa pensja Anny:", nowe_pensje['Anna'])
```

## Ćwiczenie 2: DataFrame — tworzenie i atrybuty

### Krok 1 — DataFrame z dict

```python
sklep = pd.DataFrame({
    'produkt':         ['Laptop', 'Tablet', 'Smartfon', 'Słuchawki', 'Monitor'],
    'cena':            [3500, 1800, 2500, 350, 2200],
    'stan_magazynowy': [45, 120, 200, 500, 75],
    'kategoria':       ['Komputery', 'Mobilne', 'Mobilne', 'Akcesoria', 'Komputery']
})
sklep
```

### Zadanie 2a — Eksploracja DataFrame `sklep`

```python
# 1. Kształt tabeli
print("Shape:", sklep.shape)

# 2. Nazwy kolumn
print("Kolumny:", list(sklep.columns))

# 3. Typy danych
print("\nTypy danych:")
print(sklep.dtypes)

# 4. Pierwsze 3 wiersze
print("\nPierwsze 3 wiersze:")
sklep.head(3)
```

### Zadanie 2b — Nowe kolumny

```python
# 1. Wartość magazynu
sklep['wartosc_magazynu'] = sklep['cena'] * sklep['stan_magazynowy']

# 2. Cena z VAT
sklep['cena_z_vat'] = sklep['cena'] * 1.23

# 3. Wyświetl cały DataFrame
print("DataFrame z nowymi kolumnami:")
print(sklep)

# Sprawdzenie: wartość magazynu Laptop = 3500 × 45 = 157500
print("\nWartość magazynu Laptop:", sklep.loc[sklep['produkt'] == 'Laptop', 'wartosc_magazynu'].values[0])
```

## Ćwiczenie 3: Eksploracja datasetu `penguins` — EDA

```python
url_penguins = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv'
penguins = pd.read_csv(url_penguins)
```

### Zadanie 3a — Pierwszy rzut oka

```python
# 1. Pierwsze 10 wierszy
print("head(10):")
penguins.head(10)
```

```python
# 2. Shape
print("Shape:", penguins.shape)
```

```python
# 3. info() — brakujące wartości
penguins.info()
```

### Zadanie 3b — Statystyki opisowe

```python
# 1. describe() — tylko kolumny numeryczne
print("describe():")
penguins.describe()
```

```python
# 2. describe(include='all') — wszystkie kolumny
print("describe(include='all'):")
penguins.describe(include='all')
```

```python
# 3. Mediana masy ciała
mediana_masy = penguins['body_mass_g'].median()
print("Mediana masy ciała (body_mass_g):", mediana_masy, "g")
```

### Zadanie 3c — Rozkłady wartości

```python
# 1. Pingwiny wg gatunku
print("Liczba pingwinów wg gatunku:")
print(penguins['species'].value_counts())

# 2. Pingwiny wg wyspy
print("\nLiczba pingwinów wg wyspy:")
print(penguins['island'].value_counts())

# 3. Rozkład płci
print("\nRozkład płci:")
print(penguins['sex'].value_counts())

# 4. Procent gatunku Adelie
print("\nProcent każdego gatunku:")
print(penguins['species'].value_counts(normalize=True).round(3) * 100)
```

### Zadanie 3d — Brakujące dane

```python
# 1. Brakujące wartości w każdej kolumnie
print("Brakujące wartości w każdej kolumnie:")
print(penguins.isna().sum())

# 2. Procent braków w kolumnie 'sex'
braki_sex = penguins['sex'].isna().sum()
procent_sex = braki_sex / len(penguins) * 100
print(f"\nBraki w kolumnie 'sex': {braki_sex} ({procent_sex:.1f}%)")
```

### Zadanie 3e — Wnioski z EDA

1. **Dataset zawiera 344 pingwiny** z 3 gatunków (Adelie, Chinstrap, Gentoo) obserwowanych na 3 wyspach Antarktyki.
2. **Najczęstszy gatunek to Adelie** — 152 osobniki, co stanowi ok. 44,2% wszystkich pingwinów w datasecie.
3. **Brakuje danych w 5 kolumnach** — kolumny pomiarowe (dziób, płetwa, masa) mają po 2 braki każda, natomiast kolumna `sex` ma aż 11 brakujących wartości (3,2% danych).
4. **Średnia masa ciała wynosi ok. 4202 g, a mediana 4050 g** — lekka asymetria sugeruje, że niewielka grupa ciężkich osobników (gatunek Gentoo) podciąga średnią w górę.
5. **Pingwiny żyją na 3 wyspach; najliczniej zamieszkują Biscoe** (168 osobników), a najmniej liczna jest wyspa Torgersen (52 osobniki).
6. **Rozkład płci jest zbliżony do zrównoważonego** — 165 samic i 168 samców wśród osobników z określoną płcią, co stanowi niemal idealny podział 50/50.

## Ćwiczenie 4: Pytania biznesowe — dataset `tips`

```python
# Dodaj kolumnę z procentem napiwku
tips['tip_pct'] = (tips['tip'] / tips['total_bill'] * 100).round(1)
tips.head()
```

### Zadanie 4a — Odpowiedzi na pytania biznesowe

```python
# 1. Średni rachunek
print("Średni rachunek (total_bill):", round(tips['total_bill'].mean(), 2), "USD")

# 2. Średni napiwek w procentach
print("Średni napiwek (tip_pct):", round(tips['tip_pct'].mean(), 1), "%")

# 3. Dzień z największą liczbą rachunków
print("\nLiczba rachunków wg dnia:")
print(tips['day'].value_counts())
print("Dzień z najwięcej rachunkami:", tips['day'].value_counts().idxmax())

# 4. Rachunki od palaczy
print("\nLiczba rachunków wg palacza/niepalacza:")
print(tips['smoker'].value_counts())

# 5. describe() dla 'tip' — mediana
print("\nStatystyki kolumny 'tip':")
print(tips['tip'].describe())
print("Mediana napiwku:", tips['tip'].median(), "USD")

# 6. Procent rachunków z obiadu (Dinner)
print("\nRozkład pory posiłku (%):")  
print((tips['time'].value_counts(normalize=True) * 100).round(1))

# 7. Unikalne rozmiary grup
print("\nLiczba unikalnych rozmiarów grup:", tips['size'].nunique())
print("Dostępne rozmiary:", sorted(tips['size'].unique()))
```
