Ćwiczenie 3: Operacje wektorowe — analiza biznesowa (25 min)


```python
import numpy as np

nazwy = ['Laptop', 'Tablet', 'Smartfon', 'Słuchawki', 'Ładowarka',
         'Klawiatura', 'Mysz', 'Monitor', 'Kamera', 'Drukarka']

ceny = np.array([3499, 1899, 2499, 349, 89, 249, 129, 1499, 599, 899])
ilosci_magazyn = np.array([15, 42, 78, 156, 312, 89, 201, 23, 45, 34])
oceny_klientow = np.array([4.7, 4.2, 4.5, 4.8, 3.9, 4.1, 4.3, 4.6, 3.8, 3.5])

nazwy = np.array(nazwy)
```

## Zadanie 1 — Podstawowe statystyki cenowe

```python
srednia_cena = ceny.mean()
najdrozszy_idx = ceny.argmax()
najtanszy_idx = ceny.argmin()

print(f"Średnia cena: {srednia_cena:.1f} zł")
print(f"Najdroższy produkt: {nazwy[najdrozszy_idx]} ({ceny[najdrozszy_idx]} zł)")
print(f"Najtańszy produkt: {nazwy[najtanszy_idx]} ({ceny[najtanszy_idx]} zł)")
```

## Zadanie 2 — Wartość magazynu

```python
wartosci = ceny * ilosci_magazyn
laczna_wartosc = wartosci.sum()

print("Wartość każdego produktu:", wartosci)
print(f"Łączna wartość magazynu: {laczna_wartosc} zł")
```

## Zadanie 3 — Analiza cenowa

```python
ceny_po_rabacie = ceny * 0.85
ceny_z_vat = ceny * 1.23
produkty_powyzej_1000 = (ceny > 1000).sum()

print("Ceny po rabacie 15%:", ceny_po_rabacie)
print("Ceny z VAT 23%:", ceny_z_vat)
print(f"Ile produktów kosztuje powyżej 1000 zł: {produkty_powyzej_1000}")
```

## Zadanie 4 — Filtrowanie

```python
maska_oceny = oceny_klientow >= 4.5
print("Produkty z oceną >= 4.5:")
print("Nazwy:", nazwy[maska_oceny])
print("Ceny:", ceny[maska_oceny])

maska_tanie = ceny < 200
print("\nProdukty poniżej 200 zł:")
print("Nazwy:", nazwy[maska_tanie])
print("Ilości w magazynie:", ilosci_magazyn[maska_tanie])
```

## Zadanie 5 — Analiza zapasów

```python
maska_niski_stan = ilosci_magazyn < 30

niskie_produkty = nazwy[maska_niski_stan]
wartosc_niskich = wartosci[maska_niski_stan].sum()
procent_magazynu = (wartosc_niskich / laczna_wartosc) * 100

print("Produkty z niskim stanem magazynowym (<30):", niskie_produkty)
print(f"Łączna wartość tych produktów: {wartosc_niskich} zł")
print(f"Ile % całego magazynu stanowią: {procent_magazynu:.2f}%")
```

## Zadanie 6 — Bonus: analiza cena–ocena

```python
maska_okazje = (oceny_klientow >= 4.5) & (ceny < 1000)

print("Produkty będące okazją:")
print("Nazwy:", nazwy[maska_okazje])
print("Ceny:", ceny[maska_okazje])

srednia_ge_4 = ceny[oceny_klientow >= 4.0].mean()
srednia_lt_4 = ceny[oceny_klientow < 4.0].mean()

print(f"\nŚrednia cena produktów z oceną >= 4.0: {srednia_ge_4:.2f} zł")
print(f"Średnia cena produktów z oceną < 4.0: {srednia_lt_4:.2f} zł")
```

Ćwiczenie 4: Benchmark — lista vs NumPy (15 min)


```python
import time

# Przygotowanie danych
rozmiar = 1_000_000
lista = list(range(rozmiar))
tablica = np.array(lista)

# Test 1: Pomnóż przez 2
start = time.perf_counter()
wynik_lista = [x * 2 for x in lista]
czas_lista = time.perf_counter() - start

start = time.perf_counter()
wynik_numpy = tablica * 2
czas_numpy = time.perf_counter() - start

print(f"=== Mnożenie ×2 ({rozmiar:,} elementów) ===")
print(f"Lista:  {czas_lista*1000:.1f} ms")
print(f"NumPy:  {czas_numpy*1000:.1f} ms")
print(f"NumPy jest {czas_lista/czas_numpy:.0f}× szybszy!")
```

```python
# Test 2: Suma
start = time.perf_counter()
suma_lista = sum(lista)
czas_lista = time.perf_counter() - start

start = time.perf_counter()
suma_numpy = tablica.sum()
czas_numpy = time.perf_counter() - start

print(f"\n=== Suma ({rozmiar:,} elementów) ===")
print(f"Lista:  {czas_lista*1000:.1f} ms")
print(f"NumPy:  {czas_numpy*1000:.1f} ms")
print(f"NumPy jest {czas_lista/czas_numpy:.0f}× szybszy!")
```

```python
# Test 3: Filtrowanie (elementy > 500000)
start = time.perf_counter()
filtr_lista = [x for x in lista if x > 500000]
czas_lista = time.perf_counter() - start

start = time.perf_counter()
filtr_numpy = tablica[tablica > 500000]
czas_numpy = time.perf_counter() - start

print(f"\n=== Filtrowanie > 500k ({rozmiar:,} elementów) ===")
print(f"Lista:  {czas_lista*1000:.1f} ms")
print(f"NumPy:  {czas_numpy*1000:.1f} ms")
print(f"NumPy jest {czas_lista/czas_numpy:.0f}× szybszy!")
```

## Wnioski z benchmarku

- NumPy jest **37x** szybszy w mnożeniu
- NumPy jest **6x** szybszy w sumowaniu
- NumPy jest **17x** szybszy w filtrowaniu
- Wniosek: przy dużych danych **zawsze** używaj NumPy zamiast list
